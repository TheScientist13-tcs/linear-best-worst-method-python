```python
from BestWorstMethod import solve_bwm, consistency_ratio
import pyomo.environ as pyo

## Example 1 in Rezaei (2016)
n = 5
best_idx = 1
worst_idx = 4
bo_vec = [2, 1, 4, 2, 8]
ow_vec = [4, 8, 2, 4, 1]

result = solve_bwm(n, bo_vec, ow_vec, best_idx, worst_idx)
cr_value = consistency_ratio(bo_vec[worst_idx], result['xi'])

print(f"weights: {result['weights']}")
print(f"xi: {result['xi']}")
print(f"consistency ratio: {cr_value}")
```
